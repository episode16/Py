```python
%timeit
import movielens_analysis as ml


# Загрузка данных
movies = ml.Movies("../datasets/movies.csv")
ratings = ml.Ratings("../datasets/ratings.csv")
tags = ml.Tags("../datasets/tags.csv")
links = ml.Links("../datasets/links.csv", "../datasets/movies.csv")
```

# Анализ фильмов

```python
%timeit
release_years = movies.dist_by_release()
print("Распределение фильмов по годам выпуска:", release_years)
```

```python
%timeit
genres_distribution = movies.dist_by_genres()
print("Распределение фильмов по жанрам:", genres_distribution)
```

```python
%timeit
top_genres_movies = movies.most_genres(5)
print("Фильмы с наибольшим количеством жанров:", top_genres_movies)
```

# Анализ рейтингов

```python
%timeit
ratings_by_year = ratings.Movies(ratings).dist_by_year()
print("Распределение рейтингов по годам:", ratings_by_year)
```

```python
%timeit
ratings_distribution = ratings.Movies(ratings).dist_by_rating()
print("Распределение рейтингов по оценкам:", ratings_distribution)
```

```python
%timeit
top_rated_movies = ratings.Movies(ratings).top_by_num_of_ratings(5)
print("Топ фильмов по количеству оценок:", top_rated_movies)
```

```python
%timeit
top_avg_rated_movies = ratings.Movies(ratings).top_by_ratings(5, metric="average")
print("Топ фильмов по средней оценке:", top_avg_rated_movies)
```

```python
%timeit
top_median_rated_movies = ratings.Movies(ratings).top_by_ratings(5, metric="median")
print("Топ фильмов по медианной оценке:", top_median_rated_movies)
```

```python
%timeit
top_controversial_movies = ratings.Movies(ratings).top_controversial(5)
print("Топ спорных фильмов:", top_controversial_movies)
```

```python
%timeit
dist_by_num_of_ratings_users = ratings.Users(ratings).dist_by_num_of_ratings()
print("Распределение пользователей по количеству отзывов:", dist_by_num_of_ratings_users)
```

```python
%timeit
dist_by_avg_or_median_rating_users = ratings.Users(ratings).dist_by_avg_or_median_rating(metric="median")
print("Распределение пользователей по медианной оценке:", dist_by_avg_or_median_rating_users)
```

```python
%timeit
dist_by_avg_or_median_rating_users = ratings.Users(ratings).dist_by_avg_or_median_rating(metric="average")
print("Распределение пользователей по средней оценке:", dist_by_avg_or_median_rating_users)
```

```python
%timeit
top_by_variance_users = ratings.Users(ratings).top_by_variance(5)
print("Топ пользователей с большой дисперсией оценок:", top_by_variance_users) 
```

# Анализ тегов

```python
%timeit
top_tags_by_words = tags.most_words(5)
print("Топ тегов по количеству слов:", top_tags_by_words)
```

```python
%timeit
longest_tags = tags.longest(5)
print("Самые длинные теги:", longest_tags)
```

```python
%timeit
most_popular_tags = tags.most_popular(5)
print("Самые популярные теги:", most_popular_tags)
```

```python
%timeit
tags_with_word = tags.tags_with("movie")
print("Теги, содержащие слово 'movie':", tags_with_word)
```

# Анализ ссылок

```python
%timeit

# Загрузка данных
imdb_info = links.get_imdb(default_movie_limit=9742);
links.print_imdb_info(imdb_info)
```

```python
%timeit
top_directors = links.top_directors(5, imdb_info=imdb_info)
print("Топ режиссеров:", top_directors)
```

```python
%timeit
most_expensive_movies = links.most_expensive(5, imdb_info=imdb_info)
print("Самые дорогие фильмы:", most_expensive_movies)
```

```python
%timeit
most_profitable_movies = links.most_profitable(5, imdb_info=imdb_info)
print("Самые прибыльные фильмы:", most_profitable_movies)
```

```python
%timeit
longest_movies = links.longest(5, imdb_info=imdb_info)
print("Самые длинные фильмы:", longest_movies)
```

```python
%timeit
top_cost_per_minute_movies = links.top_cost_per_minute(5, imdb_info=imdb_info)
print("Фильмы с наибольшей стоимостью за минуту:", top_cost_per_minute_movies)

```
