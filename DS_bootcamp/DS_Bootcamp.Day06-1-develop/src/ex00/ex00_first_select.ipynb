```python
import pandas as pd
import sqlite3
```

```python
cnx = sqlite3.connect('../../data/checking-logs.sqlite')
df = pd.io.sql.read_sql('PRAGMA table_info(pageviews);',cnx)
df_preview = pd.read_sql('SELECT * FROM pageviews LIMIT 10;', cnx)
print(df_preview)

query = '''
SELECT uid, datetime
FROM pageviews
WHERE uid LIKE 'user_%'
ORDER BY uid ASC
'''
df_pageviews = pd.read_sql(query, cnx)
print(df_pageviews)

df_pageviews['datetime'] = pd.DatetimeIndex(df_pageviews['datetime'])
df_pageviews.set_index('datetime', inplace=True)
print(df_pageviews)

cnx.close()
print(df_pageviews.head())
```
