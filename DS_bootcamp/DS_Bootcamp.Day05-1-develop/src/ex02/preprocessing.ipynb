```python
import pandas as pd
```

```python
df = pd.read_csv('../../data/auto.csv', sep = ',')
df.set_index('ID', inplace=True)
print(df)
print('before deletion duplicates:\n', df.count())
```

```python
df = df.drop_duplicates(subset=['CarNumber', 'Make_n_model', 'Fines'], keep='last')
print('after deletion duplicates:\n', df.count())
```

```python
count_of_nan_values = df.isna().sum()
print('count of nan values:\n',count_of_nan_values)
```

```python
df = df.dropna(axis=1, thresh=500)
print(df)
```

```python
if 'Refund' in df.columns:
    df['Refund'] = df['Refund'].ffill()
print("После заполнения 'Refund':\n", df.isna().sum())
```

```python
if 'Fines' in df.columns:
    mean_value = df['Fines'].mean()
    df['Fines'] = df['Fines'].fillna(mean_value)

print("После заполнения 'Fines':\n", df.isna().sum())
```

```python
df['Make'] = df['Make_n_model'].apply(lambda x: x.split()[0] if isinstance(x, str) else x)
df['Model'] = df['Make_n_model'].apply(lambda x: x.split()[1] if isinstance(x, str) and len(x.split()) > 1 else None)
df
```

```python
df = df.drop('Make_n_model', axis=1)
```

```python
df.to_json (r'../../data/auto.json')
```
