```python
import pandas as pd
```

```python
df = pd.read_json('../../data/auto.json')
df.set_index('CarNumber', inplace=True)
```

```python
filter_fines_2100 = df[df['Fines'] > 2100]
filter_fines_2100
```

```python
filter_fines_n_refund = df[(df['Fines'] > 2100) & (df['Refund'] == 2)]
filter_fines_n_refund
```

```python
model_list = ['Focus', 'Corolla']
filter_model = df[df['Model'].isin(model_list)]
filter_model
```

```python
CarNumber_list = ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']
filter_CarNumber = df.loc[CarNumber_list]
filter_CarNumber
```

```python
median_fines_by_make = df.groupby('Make')['Fines'].median()
print(median_fines_by_make)
```

```python
median_fines_by_make_model = df.groupby(['Make', 'Model'])['Fines'].median()
print(median_fines_by_make_model)
```

```python
count_fines_by_make_model = df.groupby(['Make', 'Model'])['Fines'].count()
print(count_fines_by_make_model)
```

```python
min_max_fines_by_make_model = df.groupby(['Make', 'Model'])['Fines'].agg(['min', 'max'])
print(min_max_fines_by_make_model)
```

```python
std_fines_by_make_model = df.groupby(['Make', 'Model'])['Fines'].std()
print(std_fines_by_make_model)
```

```python
fines_count_by_car = df.groupby('CarNumber')['Fines'].count().sort_values(ascending=False)
print(fines_count_by_car)
```

```python
top_violator = fines_count_by_car.index[0]
top_violator_rows = df.loc[top_violator]
print(top_violator_rows)
```

```python
fines_sum_by_car = df.groupby('CarNumber')['Fines'].sum().sort_values(ascending=False)
print(fines_sum_by_car)
```

```python
top_payer = fines_sum_by_car.index[0]
top_payer_rows = df.loc[top_payer]
print(top_payer_rows)
```

```python
car_model_variants = df.groupby('CarNumber')['Model'].apply(lambda x: set(x))
multiple_models = car_model_variants[car_model_variants.apply(len) > 1]
print(multiple_models)
```
