```python
import pandas as pd
```

```python
df = pd.read_csv('../../data/feed-views.log', sep = '\t', names = ['datetime', 'user'])
```

```python
df['datetime'] = df['datetime'].astype("datetime64[ns]")

df['datetime'] = df['datetime'].dt.round('s') 
df['year'] = df['datetime'].dt.year
df['month'] = df['datetime'].dt.month
df['day'] = df['datetime'].dt.day
df['hour'] = df['datetime'].dt.hour
df['minute'] = df['datetime'].dt.minute
df['second'] = df['datetime'].dt.second
df
```

```python
df['daytime'] = pd.cut(df['hour'], bins=[0, 4, 7, 11, 17, 20, 24],
                    labels=['night', 'early morning', 'morning', 'afternoon', 'early evening', 'evening'],
                    right=False)
df.set_index('user')
```

```python
total_elements = df.count()
print(total_elements)

daytime_counts = df['daytime'].value_counts()
daytime_counts
```

```python
sorted_df = df.sort_values(by=['hour','minute','second'], ascending=True)
sorted_df
```

```python
max_hour_night = df[df['daytime'] == 'night']['hour'].max()
print('max night hour: ', max_hour_night)

min_hour_morning = df[df['daytime'] == 'morning']['hour'].min()
print('min morning hour: ', min_hour_morning)

user_at_max_night_hour = df[df['hour'] == max_hour_night].index[0]
print('user at max night hour: ', user_at_max_night_hour)

user_at_min_morning_hour = df[df['hour'] == min_hour_morning].index[0]
print('user at min morning hour: ', user_at_min_morning_hour)

mode_daytime = df['daytime'].mode().iat[0]
print('mode daytime: ', mode_daytime)

mode_hour = df['hour'].mode().iat[0]
print('mode hour: ', mode_hour)
```

```python
early_morning = df[df['daytime'] == 'morning'].nsmallest(3, 'hour')
print('earliest hours in the morning usernames:\n', early_morning[['user', 'hour']])

late_night = df[df['daytime'] == 'evening'].nlargest(3, 'hour')
print('\n\nlatest hours in the night usernames:\n',late_night[['user', 'hour']])
```
