```python
import pandas as pd
import numpy as np
```

```python
pd.options.display.float_format = '{:,.2f}'.format
df = pd.read_json('../../data/auto.json')
df
```

```python
sample = df[['CarNumber', 'Make', 'Model']].sample(n=200, replace=True, random_state=21)
sample['Fines'] = np.random.choice(df['Fines'], size=200, replace=True)
sample['Refund'] = np.random.choice(df['Refund'], size=200, replace=True)
concat_rows = pd.concat([df, sample], ignore_index=True)
concat_rows
```

```python
np.random.seed(21)
'''concat_rows['Year'] = np.random.randint(1980, 2020, size=len(concat_rows))
concat_rows'''
year_series = pd.Series(np.random.randint(1980, 2020, size=len(concat_rows)), name='Year')
fines = pd.concat([concat_rows, year_series], axis=1)
fines
```

```python
df_names = pd.read_json('../../datasets/surname.json')
df_names.columns = df_names.iloc[0]
df_names = df_names[1:].reset_index(drop=True)

surnames = df_names['NAME'].tolist()

np.random.seed(21)
unique_cars = df['CarNumber'].unique()
random_surnames = np.random.choice(surnames, size=len(unique_cars), replace=True)

owners = pd.DataFrame({'CarNumber': unique_cars, 'SURNAME': random_surnames})

new_fines = pd.DataFrame({
    'CarNumber': ['NEW001', 'NEW002', 'NEW003', 'NEW004', 'NEW005'],
    'Make': ['Toyota', 'Ford', 'Honda', 'BMW', 'Audi'],
    'Model': ['Camry', 'Focus', 'Civic', 'X5', 'A4'],
    'Fines': [500, 1500, 1200, 2000, 1700],
    'Refund': [1, 2, 0, 3, 2],
    'Year': [2010, 2012, 2015, 2018, 2019]
})

fines = pd.concat([fines, new_fines], ignore_index=True)
owners = owners.iloc[:-20]

new_owners = pd.DataFrame({
    'CarNumber': ['UNIQUE01', 'UNIQUE02', 'UNIQUE03'],
    'SURNAME': ['Smith', 'Johnson', 'Williams']
})

owners = pd.concat([owners, new_owners], ignore_index=True)
fines.set_index('CarNumber', inplace=True)
owners.set_index('CarNumber', inplace=True)

df_1 = fines.join(owners, how='inner')
df_2 = fines.join(owners, how='outer')
df_3 = fines.join(owners, how='left')
df_4 = fines.join(owners, how='right')

pivot_table = fines.pivot_table(index='Make', columns='Year', values='Fines', aggfunc='sum')
pivot_table
```

```python
owners.to_csv('../../data/owners.csv', index=True)
fines.to_csv('../../data/fines.csv', index=True) 
```
