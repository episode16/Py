```python
import pandas as pd
import sqlite3
```

```python
cnx = sqlite3.connect('../../datasets/checking-logs.sqlite')

query_checker = '''
SELECT uid, timestamp FROM checker
WHERE uid NOT LIKE '%admin%'
'''
df_checker = pd.read_sql(query_checker, cnx)

df_checker['datetime'] = pd.to_datetime(df_checker['timestamp'])

df_checker['datetime'] = df_checker['datetime'].dt.round('s') 
df_checker['year'] = df_checker['datetime'].dt.year
df_checker['month'] = df_checker['datetime'].dt.month
df_checker['day'] = df_checker['datetime'].dt.day
df_checker['hour'] = df_checker['datetime'].dt.hour
df_checker['minute'] = df_checker['datetime'].dt.minute
df_checker['second'] = df_checker['datetime'].dt.second
df_checker['daytime'] = pd.cut(df_checker['hour'], bins=[0, 4, 7, 11, 17, 20, 24],
                               labels=['night', 'early morning', 'morning', 'afternoon', 'early evening', 'evening'],
                               right=False)
df_checker
```

```python
df_grouped = df_checker.groupby([df_checker['datetime'].dt.date, 'daytime'], observed=False).size().unstack(fill_value=0)

ax = df_grouped.plot(kind='bar', stacked=True, figsize=(15, 8))
ax.legend(title='daytime', fontsize = 8)
ax.set_xlabel('date', fontsize = 8)
ax.set_title('Commits per day', fontsize = 8)
```

```python
max_commits_in_daytime = df_checker['daytime'].value_counts().head(2)
print('Most common periods for committing labs:', max_commits_in_daytime)
```

```python
df_day_commits = df_checker.groupby([df_checker['datetime'].dt.date, 'daytime'], observed=False).size().unstack(fill_value=0)
valid_days = df_day_commits[df_day_commits['evening'] > df_day_commits['afternoon']]
day_with_max_commits = valid_days.sum(axis=1).idxmax()

print("Day with most commits and evening > afternoon:", day_with_max_commits)
cnx.close()
```
