```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
```

```python
cnx = sqlite3.connect('../../datasets/checking-logs.sqlite')
query = '''
SELECT uid, timestamp FROM checker
WHERE uid NOT LIKE '%admin%'
'''
df = pd.read_sql(query, cnx)
cnx.close()
df['datetime'] = pd.to_datetime(df['timestamp'])
df ['hour']  = df['datetime'].dt.hour
df['weekday'] = df['datetime'].dt.weekday
df
```

```python
df['weekends'] = df['weekday'] >= 5
df['workdays'] = df['weekday'] < 5
weekend_hours = df.loc[df['weekends'], 'hour'].tolist()
workday_hours = df.loc[df['workdays'], 'hour'].tolist()
```

```python
plt.figure(figsize=(15, 8))

plt.hist(workday_hours, bins=24, color='blue', alpha=0.7, label='Working days')

plt.hist(weekend_hours, bins=24, color='red', alpha=0.7, label='Weekends')
plt.xlim(0, 23)
plt.xlabel('Hour', fontsize=8)
plt.ylabel('Number of commits', fontsize=8)
plt.title('Commit distribution per hour', fontsize=8, fontweight='bold')
plt.legend()


plt.show()
```

Are there hours when the total number of commits was higher on weekends than
on working days?

```python
workday_counts = df.loc[df['workdays'], 'hour'].value_counts().sort_index()
weekend_counts = df.loc[df['weekends'], 'hour'].value_counts().sort_index()

result = []

for hour in range(24):
    workday_commits = workday_counts.get(hour, 0)
    weekend_commits = weekend_counts.get(hour, 0)
    
    if weekend_commits > workday_commits and len(result) < 4:
        result.append(hour)

print(*result)
```
