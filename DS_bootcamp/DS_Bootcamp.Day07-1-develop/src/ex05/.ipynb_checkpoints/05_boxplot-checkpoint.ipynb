```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../datasets/ab-test.csv')
cnx = sqlite3.connect('ab-test.db')
df.to_sql('ab_test', cnx, index=False)
df_from_db = pd.read_sql("SELECT * FROM ab_test;", cnx)
print(df_from_db)
```

```python
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(15, 8))
group_after_control = df.loc[(df['group'] == 'control') & (df['time'] == 'after'), 'diff']
group_after_test = df.loc[(df['group'] == 'test') & (df['time'] == 'after'), 'diff']
group_before_control = df.loc[(df['group'] == 'control') & (df['time'] == 'before'), 'diff']
group_before_test = df.loc[(df['group'] == 'test') & (df['time'] == 'before'), 'diff']

boxprops = dict(color='green', linewidth=3)
medianprops = dict(color='red', linewidth=2)
whiskerprops = dict(color='red')
capprops = dict(color='red')

axs[0].boxplot([group_after_control, group_after_test], 
               tick_labels=['control', 'test'],
               boxprops=boxprops, 
               medianprops=medianprops, 
               whiskerprops=whiskerprops, 
               capprops=capprops)

axs[0].set_title('after', fontsize=14)
axs[0].set_ylabel('diff')
axs[0].grid()

axs[1].boxplot([group_before_control, group_before_test], 
               tick_labels=['control', 'test'],
               boxprops=boxprops, 
               medianprops=medianprops, 
               whiskerprops=whiskerprops, 
               capprops=capprops)

axs[1].set_title('before', fontsize=14)
axs[1].set_ylabel('diff')
axs[1].grid()
plt.suptitle('Lag before labs deadlines. After and before the Newsfeed', fontsize=14)
plt.show()
```

What was the IQR of the control group before the newsfeed?

```python
Q1 = group_before_control.quantile(0.25)
Q3 = group_before_control.quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```
