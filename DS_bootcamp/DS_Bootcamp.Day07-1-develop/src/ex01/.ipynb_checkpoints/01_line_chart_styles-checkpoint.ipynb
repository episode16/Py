```python
import pandas as pd
import sqlite3
```

1

```python
cnx = sqlite3.connect('../../datasets/checking-logs.sqlite')
#For views
query_views = '''
SELECT uid, datetime FROM pageviews
WHERE uid NOT LIKE '%admin%'
'''
df_views = pd.read_sql(query_views, cnx)
df_views["datetime"] = pd.to_datetime(df_views["datetime"])
df_views["date"] = df_views["datetime"].dt.date

df_grouped_views = df_views.groupby("date").size().reset_index(name="views")

#For checker
df = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", cnx)
#print(df)
query_checker = '''
SELECT uid, timestamp FROM checker
WHERE uid NOT LIKE '%admin%'
'''
df_checker = pd.read_sql(query_checker, cnx)
df_checker["datetime"] = pd.to_datetime(df_checker["timestamp"])
df_checker["date"] = df_checker["datetime"].dt.date

df_grouped_checker = df_checker.groupby("date").size().reset_index(name="views")
print(df_grouped_views)
print(df_grouped_checker)
```

```python
df_merged = pd.merge(df_grouped_views, df_grouped_checker, on="date", how="outer")
print(df_merged)
df_merged.rename(columns={'views_x': 'views', 'views_y': 'commits'}, inplace=True)

graph = df_merged.plot(x="date", y="views", figsize=(15, 8), title="Views and Commits per day", fontsize=8,
                    linestyle="-.", marker="o", color="red", label="Views")
df_merged.plot(x="date", y="commits", ax=graph, linestyle="--", color="blue", label="Commits", fontsize = 8)

graph.set_xlabel('date', fontsize = 8)
```

```python
count_views_above_150 = df_merged[df_merged['views'] > 150].shape[0]
print('How many times was the number of views larger than 150? The answer is', count_views_above_150)
```
