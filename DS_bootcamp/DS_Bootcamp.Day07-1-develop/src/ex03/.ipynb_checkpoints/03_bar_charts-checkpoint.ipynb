```python
import pandas as pd
import sqlite3
```

```python
cnx = sqlite3.connect('../../datasets/checking-logs.sqlite')
query = '''
SELECT uid, timestamp FROM checker
WHERE uid NOT LIKE '%admin%'
'''
df = pd.read_sql(query, cnx)
df['datetime'] = pd.to_datetime(df['timestamp'])
df ['hour']  = df['datetime'].dt.hour
df['weekday'] = df['datetime'].dt.weekday
```

```python
df['weekends'] = df['weekday'] >= 5
avg_commits = df.groupby(['hour', 'weekends'])['uid'].nunique().reset_index()
avg_commits = avg_commits.rename(columns={'uid': 'avg_commits'})

all_hours = pd.DataFrame({'hour': range(24)})
weekend_commits = avg_commits[avg_commits['weekends'] == True]
weekend_commits = all_hours.merge(weekend_commits, on='hour', how='left')

workday_commits = avg_commits[avg_commits['weekends'] == False]
workday_commits = all_hours.merge(workday_commits, on='hour', how='left')
```

```python
ax_1 = weekend_commits.plot(x='hour', y='avg_commits', kind='bar', color='lightgreen', figsize=(15, 8), legend=True, label='weekend')
ax_2 = workday_commits.plot(x='hour', y='avg_commits', kind='bar', color='gray', figsize=(15, 8), legend=True, label='working_day')
ax_1.annotate('Commits per hour', xy=(0.5, 1.05), xycoords='axes fraction', ha='center', va='bottom', fontsize=14, fontweight='bold')
ax_1.legend(title='daytime', fontsize = 8)
ax_1.set_xlabel('hour', fontsize = 8)
ax_1.set_title('Weekend', fontsize = 8)

ax_2.legend(title='daytime', fontsize = 8)
ax_2.set_xlabel('hour', fontsize = 8)
ax_2.set_title('Working day', fontsize = 8)

ax_1.tick_params(axis='x', rotation=0)
ax_2.tick_params(axis='x', rotation=0)

```

“Is the dynamic different on working days and weekends?”

```python
max_workday_hour = workday_commits.loc[workday_commits['avg_commits'].idxmax(), 'hour']
max_weekend_hour = weekend_commits.loc[weekend_commits['avg_commits'].idxmax(), 'hour']

print(f"The peak number of commits during working days is at {max_workday_hour}:00.  \n")
print(f"The peak number of commits during weekends is at {max_weekend_hour}:00.")
```
