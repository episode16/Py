```python
def search_and_draw(X, y, model, param_name, grid, param_scale='ordinary', draw=True):
    """
    Универсальная функция для поиска параметров и визуализации результатов

    Параметры:
    ----------
    X : array-like of shape (n_samples, n_features)
        Матрица признаков
    y : array-like of shape (n_samples,)
        Целевая переменная
    model : estimator object или строка ('LogisticRegression', 'LogisticRegressionCV')
        Модель для тестирования
    param_name : str
        Название параметра для поиска ('C' или 'Cs')
    grid : array-like
        Сетка значений параметра
    param_scale : str ('ordinary' или 'log')
        Тип шкалы для оси X
    draw : bool
        Флаг отображения графика

    Возвращает:
    -----------
    dict: Словарь с результатами (лучшие параметры, метрики и т.д.)
    """
    if isinstance(model, str):
        if model == 'LogisticRegression':
            model = LogisticRegression(max_iter=1000, class_weight='balanced')
        elif model == 'LogisticRegressionCV':
            model = LogisticRegressionCV(cv=5, scoring='roc_auc', max_iter=1000)
            param_scale = 'log'  # Для LogisticRegressionCV всегда используем log scale

    results = {}

    # Для LogisticRegressionCV используем встроенную кросс-валидацию
    if isinstance(model, LogisticRegressionCV):
        model.set_params(Cs=grid)
        model.fit(X, y)

        best_param = model.C_[0]
        best_score = model.scores_[1].mean(axis=0).max()
        means = model.scores_[1].mean(axis=0)
        errors = model.scores_[1].std(axis=0)

        results = {
            'best_param': best_param,
            'best_score': best_score,
            'means': means,
            'errors': errors,
            'model': model
        }

    # Для остальных моделей используем GridSearchCV
    else:
        CV_model = GridSearchCV(
            estimator=model,
            param_grid={param_name: grid},
            cv=5,
            scoring='roc_auc',
            n_jobs=-1,
            verbose=0
        )
        CV_model.fit(X, y)

        best_param = CV_model.best_params_[param_name]
        best_score = CV_model.best_score_
        means = CV_model.cv_results_['mean_test_score']
        errors = CV_model.cv_results_['std_test_score']

        results = {
            'best_param': best_param,
            'best_score': best_score,
            'means': means,
            'errors': errors,
            'model': CV_model.best_estimator_
        }

    print("\n" + "="*50)
    print(f"Лучшая модель: {model.__class__.__name__}")
    print(f"Лучший параметр {param_name}: {best_param}")
    print(f"Лучший ROC-AUC: {best_score:.4f}")
    print("="*50 + "\n")

    if draw:
        plt.figure(figsize=(12, 6))
        plt.title(f'Оптимизация параметра {param_name} ({model.__class__.__name__})')

        if param_scale == 'log':
            plt.xscale('log')

        plt.plot(grid, means, 'o-', label='Средний ROC-AUC', color='blue', lw=2)
        plt.scatter(best_param, best_score, color='red', s=150,
                   label=f'Лучший: {param_name}={best_param:.3f}\nROC-AUC={best_score:.3f}')

        plt.fill_between(grid, means - errors, means + errors,
                        color='blue', alpha=0.2, label='±1σ доверительный интервал')

        plt.xlabel(f'Значение параметра {param_name}')
        plt.ylabel('ROC-AUC')
        plt.grid(True, which='both', ls='--', alpha=0.5)
        plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
        plt.tight_layout()
        plt.show()

    return results
```

```python
models = [LogisticRegression(), LogisticRegressionCV()]
param_names = ['C', 'Cs']
grids = [
    [100, 10, 1, 0.1, 0.01, 0.001],
    [100, 10, 1, 0.1, 0.01, 0.001]
]
param_scales = ['log', 'ordinary']
```

```python
for model, param_name, grid, param_scale in zip(models,
                                                param_names,
                                                grids,
                                                param_scales):
    search_and_draw(X_train, y_train, model, param_name, grid, param_scale)
```
